```python
# Remove null values
df = df.dropna()

# Convert date column (if exists)
df['date'] = pd.to_datetime(df['date'])

# Remove duplicates
df = df.drop_duplicates()
```

```python
df
```

```python
df.isnull().sum()
```

```python
df['month'] = df['date'].dt.month
df['year'] = df['date'].dt.year
```

```python
Total_revenue  = df['sales_revenue_usd'].sum()
print("Total sales revenue:", Total_revenue)
```

```python
df.groupby('region')['sales_revenue_usd'].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt

df.groupby('region')['sales_revenue_usd'].sum().plot(kind='bar')
plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.show()
```

```python
import matplotlib.pyplot as plt

df.groupby('region')['sales_revenue_usd'].sum().plot(kind='line')
plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.show()
```

```python
import matplotlib.pyplot as plt

df.groupby('region')['sales_revenue_usd'].sum().plot(kind='pie')
plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.show()
```
